a. Write a Python program to create a scatter plot of Age and Height of 5
students.

```python
import matplotlib.pyplot as plt

age = [12, 13, 14, 15, 16]
height = [145, 150, 155, 160, 165]

plt.scatter(age, height, color='red')

plt.title("Age and Height of Students")
plt.xlabel("Age")
plt.ylabel("Height (cm)")
plt.show()
```

b. Write a Python program to create a bar chart showing marks of 5
students.

```python
import matplotlib.pyplot as plt

students = ["A", "B", "C", "D", "E"]
marks = [85, 72, 90, 65, 78]

plt.bar(students, marks, color='green')

plt.title("Marks of 5 Students")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

c. Write a Python program to create a histogram showing the distribution of
marks of 10 students.

```python
import matplotlib.pyplot as plt

marks = [45, 56, 67, 72, 81, 65, 90, 55, 76, 88]

plt.hist(marks, bins=5, color='blue', edgecolor='black')

plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
```

d. Write a Python program to create a pie chart showing the number of
students in different courses.

```python
import matplotlib.pyplot as plt

courses = ["BCA", "BBA", "B.Sc", "BA"]
students = [30, 25, 20, 15]

plt.pie(students, labels=courses, autopct='%1.1f%%')

plt.title("Students in Different Courses")
plt.show()
```

e. Write a Python program to create a bar chart showing sales of 5 different
products.

```python
import matplotlib.pyplot as plt

products = ["Product A", "Product B", "Product C", "Product D", "Product E"]
sales = [120, 150, 100, 180, 140]

plt.bar(products, sales, color='pink')

plt.title("Sales of Different Products")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```
